## Missing Data Handling

### Missing Data

```r
x = NA
is.na(x)
```

### How to work with missing data

```r
x = c(11, NA, 13, NA)
mean(x)
```

```r
mean(x, na.rm = TRUE)
```

### Location of missing values

```r
x = c(11, NA, 13, NA)
x
```

```r
which(is.na(x))
```

### Counting number of missing values

```r
x = c(11, NA, 13, NA)
x
```

```r
sum(is.na(x))
```

### Finding complete cased

```r
x = c(11, NA, 13, NA)
x
```

```r
complete.cases(x)
```

```r
x = c(11, NA, 13, NA)
x
```

```r
y = na.omit(x)
y
```

```r
mean(x)
```

```r
mean(y)
```

#### Example 1

```r
x = 5
if(x==3){
  x = x - 1
} else{
  x = 2*x
}
x
```

```r
x = 3
if(x==3){
  x = x - 1
} else{
  x = 2*x
}
x
```
