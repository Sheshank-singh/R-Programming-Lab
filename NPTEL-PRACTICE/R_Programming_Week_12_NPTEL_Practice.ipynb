## Subdivided or Component Bar Plots

A component bar plot shows totals divided into parts. In R, a matrix is commonly used: each column forms a bar and the segments represent the component values.

```r
cust <- matrix(
  nrow = 4, ncol = 3,
  data = c(2, 20, 30, 26, 53, 40, 42, 15, 25, 30, 75, 100),
  byrow = TRUE
)
cust
```

```r
barplot(cust)
```

### Adding labels and colours

```r
barplot(
  cust,
  names.arg = c("Shop 1", "Shop 2", "Shop 3"),
  xlab = "Shops",
  ylab = "Customers",
  col = c("red", "green", "orange", "brown"),
  main = "Customers visiting three shops"
)
```

## Combining Multiple Graphics

Use `par(mfrow = c(rows, columns))` to arrange multiple plots in one graphics window.

```r
par(mfrow = c(1, 2))
barplot(table(direction), main = "Bar Plot")
pie(table(direction), main = "Pie Chart")
```

```r
par(mfrow = c(2, 1))
barplot(table(direction), main = "Bar Plot")
pie(table(direction), main = "Pie Chart")
```

```r
par(mfrow = c(1, 1))  # Reset to one plot per graphics window
```

## Histograms

A histogram summarizes a continuous numeric variable by grouping values into intervals (bins). The bar area represents frequency; `freq = FALSE` displays a density scale.

```r
height <- c(
  166,125,130,142,147,159,159,147,165,156,149,164,137,166,135,142,
  133,136,127,143,165,121,142,148,158,146,154,157,124,125,158,159,
  164,143,154,152,141,164,131,152,152,161,143,143,139,131,125,145,
  140,163
)
length(height)
```

```r
hist(height)
```

```r
hist(
  height,
  main = "Heights of Persons",
  col = "green",
  xlab = "Height (cm)",
  ylab = "Number of Persons"
)
```

```r
hist(
  height,
  main = "Heights of Persons",
  col = "red",
  xlab = "Height (cm)",
  ylab = "Density",
  freq = FALSE
)
```

```r
hist(
  height,
  main = "Heights of Persons",
  col = "red",
  xlab = "Height (cm)",
  ylab = "Number of Persons",
  density = 2,
  angle = 100
)
```

## Bivariate Scatter Plots

A bivariate scatter plot displays paired observations of two variables. It can help reveal positive or negative association and whether the relationship looks linear, nonlinear, or unclear.

### Marks and weekly study hours

```r
marks <- c(
  337,316,327,340,374,330,352,353,370,380,
  384,398,413,428,430,438,439,479,460,450
)
hours <- c(
  23,25,26,27,30,26,29,32,33,34,
  35,38,39,42,43,44,45,46,44,41
)
length(marks)
length(hours)
```

```r
plot(hours, marks)
```

### Different plot types

The `type` argument changes how points are displayed: `p` points, `l` lines, `b` both points and lines, `c` lines only from `b`, `o` overplotted points and lines, `s` stair steps, and `h` vertical lines.

```r
plot(hours, marks, type = "p", main = "Points")
```

```r
plot(hours, marks, type = "l", main = "Lines")
```

```r
plot(hours, marks, type = "b", main = "Points and Lines")
```

```r
plot(hours, marks, type = "c", main = "Lines Only")
```

```r
plot(hours, marks, type = "o", main = "Overplotted")
```

```r
plot(hours, marks, type = "s", main = "Stair Steps")
```

```r
plot(hours, marks, type = "h", main = "Vertical Lines")
```

### Adding titles and axis labels

```r
plot(
  hours, marks,
  xlab = "Number of weekly hours",
  ylab = "Marks obtained",
  main = "Marks obtained versus Number of hours per week"
)
```

## Matrix Scatter Plot with `pairs()`

`pairs()` creates a matrix of scatter plots for the variables supplied. It is useful for quickly inspecting relationships between multiple numeric variables.

```r
pairs(cbind(hours, marks))
```

```r
pairs(
  cbind(hours, marks),
  labels = c("Study hours", "Marks obtained"),
  col = "red"
)
```

## Scatter Plot with a Smooth Curve

`scatter.smooth()` adds a locally weighted smooth curve (LOESS) to help visualize a possible trend.

```r
scatter.smooth(hours, marks)
```

```r
scatter.smooth(
  hours, marks,
  xlab = "Weekly study hours",
  ylab = "Marks obtained",
  main = "Study hours and marks",
  lpars = list(col = "red", lwd = 3, lty = 3)
)
```
