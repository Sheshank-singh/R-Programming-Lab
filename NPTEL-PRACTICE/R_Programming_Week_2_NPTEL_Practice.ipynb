## Addition with Data Vectors

R performs element-wise addition between data vectors.

```r
c(2,3,5,7) + c(-2,-3,-5,8)
```

### Addition with vectors of different lengths

```r
c(2,3,5,7) + c(8,9)
```

### When the longer vector length is not a multiple of the shorter vector length

```r
c(2,3,5,7) + c(8,9,10)
```

## Subtraction with Data Vectors

```r
c(2,3,5,7) - c(-2,-3,-5,8)
```

### Subtraction with vectors of different lengths

```r
c(12,13,15,17) - c(8,9)
```

```r
c(12,13,15,17) - c(8,9,10)
```

## BODMAS Rule

BODMAS is applicable in R.

- B — Brackets
- O — Orders (powers)
- D — Division
- M — Multiplication
- A — Addition
- S — Subtraction

Mathematical expressions with multiple operators are solved from left to right according to this order.

Only `( )` brackets are used for BODMAS.

```r
(2+3)*5 + 5 - 10
```

```r
(((2+3)*5 + 5) - 10)/2
```
