## Packages and Libraries in R

## Libraries in R

R packages are collections of programs and datasets developed by different users and scientists.

Functions and datasets are organised into libraries.

To use a library, the `library()` function is used.

```r
library(spatial)
```

## Description of Libraries

The `packageDescription()` command provides the description file of a package.

```r
packageDescription("spatial")
```

## Contents of Libraries

The `help` function can be used to discover the contents of library packages

```r
library(help=spatial)
```

## Installing Packages and Libraries

The base R package contains programs for basic operations.

Specific requirements can be met by installing special packages.

## Installing Packages

The `install.packages()` function is used to download packages

```r
install.packages("boot")
```

```r
install.packages("cluster")
```

## Checking Installed Packages

The `installed.packages()` command is used to see the packages installed on the computer.

```r
installed.packages()
```

## Removing Installed Packages

The `remove.packages()` command is used to remove an installed package.

```r
remove.packages("cluster")
```

## Updating Installed Packages

The `update.packages()` command is used to update installed packages.

```r
update.packages("cluster")
```

## Demonstration of `persp`

The notes demonstrate the `persp` function for 3D surface plots.

```r
require(datasets)
```

```r
require(grDevices); require(graphics)
```

```r
x <- seq(-10, 10, length.out = 50)
```

```r
y <- x
```

```r
rotinc <- function(x,y)
{
    sinc <- function(x) { y <- sin(x)/x ; y[is.na(y)] <- 1; y }
    10 * sinc( sqrt(x^2+y^2) )
}
```

```r
sinc.exp <- expression(z == Sinc(sqrt(x^2 + y^2)))
```

```r
z <- outer(x, y, rotinc)
```

```r
oldpar <- par(bg = "white")
```

```r
persp(x, y, z, theta = 30, phi = 30, expand = 0.5, col = "lightblue")
```
